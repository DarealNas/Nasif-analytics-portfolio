```python
# Q4 Write a python function called check_number (num) that:
#	Prints “Positive” if the number is greater than 0
#	Prints “Negative” if lesser than 0
#	Prints “Zero” if equal to 0
```

```python
def check_number(num):
    if num > 0:
        print("Positive")
    elif num < 0:
        print("Negative")
    else:
        print("Zero")
```

```python
check_number(5)    
```

```python
check_number(-3)  
```

```python
check_number(0)  
```

```python
# Q5: Create a list of 5 fruits.
#	Add one more fruit to the list
#	Remove one fruit
#	Print the first and last fruits in the list
```

```python
# 1. Create a list of 5 fruits
fruits = ["apple", "banana", "orange", "mango", "grape"]

# 2. Add one more fruit
fruits.append("pineapple")

# 3. Remove one fruit (say "orange")
fruits.remove("orange")

# 4. Print first and last fruit
print("First fruit:", fruits[0])
print("Last fruit:", fruits[-1])
```

```python
# Q6: Create two sets:
# Find and Print 
#	The Union
#	The Intersection
#	The difference (A - B)
```

```python
# Create two sets:
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

# Union: elements in A or B
union_set = A.union(B)

# Intersection: elements in both A and B
intersection_set = A.intersection(B)

# Difference: elements in A but not in B
difference_set = A.difference(B)

print("Union:", union_set)
print("Intersection:", intersection_set)
print("Difference (A - B):", difference_set)
```
